 ## 6. Test NumPy and SciPy

```python
import numpy as np
from scipy import signal

# Generate a simple test signal
sampling_rate = 1000
time = np.arange(0, 1, 1 / sampling_rate)

signal_data = (
    np.sin(2 * np.pi * 10 * time)
    + 0.5 * np.sin(2 * np.pi * 50 * time)
)

# Design a low-pass filter
b, a = signal.butter(
    3,
    20,
    btype="low",
    fs=sampling_rate
)

filtered_signal = signal.filtfilt(b, a, signal_data)

print("Signal-processing test successful.")
print(f"Number of samples: {len(signal_data)}")
print(f"Original SD:        {np.std(signal_data):.3f}")
print(f"Filtered SD:        {np.std(filtered_signal):.3f}")
```

 ## 7. Test plotting

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 4))

ax.plot(
    time[:500],
    signal_data[:500],
    label="Original",
    alpha=0.6
)

ax.plot(
    time[:500],
    filtered_signal[:500],
    label="Filtered",
    linewidth=2
)

ax.set_xlabel("Time (s)")
ax.set_ylabel("Amplitude")
ax.set_title("Neuropixels Course Installation Test")
ax.legend()

plt.tight_layout()
plt.show()
```

 You should see a plot containing the original and filtered signals.

 ## 9. Test scikit-learn

```python
from sklearn.decomposition import PCA

# Create a small artificial population activity matrix
rng = np.random.default_rng(42)

population_activity = rng.normal(
    size=(100, 20)
)

pca = PCA(n_components=3)

reduced_activity = pca.fit_transform(
    population_activity
)

print("scikit-learn test successful.")
print(f"Original dimensions: {population_activity.shape}")
print(f"PCA dimensions:      {reduced_activity.shape}")
print(
    "Explained variance:  "
    f"{pca.explained_variance_ratio_.round(3)}"
)
```

```python
import bombcell as bc
from bc.ccg_fast import ccg_bz


"""
Test the CCG implementation with simple data
"""
# Generate test data
np.random.seed(42)

# Two units with different firing patterns
times1 = np.sort(np.random.exponential(0.1, 100))
times2 = np.sort(np.random.exponential(0.15, 80)) + 0.002  # slight offset

times = np.concatenate([times1, times2])
groups = np.concatenate([np.ones(len(times1)), np.full(len(times2), 2)]).astype(np.uint32)

# Compute CCG
ccg, t = ccg_bz(times, groups, bin_size=0.001, duration=0.1)

print(f"CCG shape: {ccg.shape}")
print(f"Time lags: {len(t)} bins from {t[0]:.3f} to {t[-1]:.3f} s")
print(f"Peak auto-correlation unit 1: {np.max(ccg[:, 0, 0])}")
print(f"Peak auto-correlation unit 2: {np.max(ccg[:, 1, 1])}")
print(f"Max cross-correlation: {np.max(ccg[:, 0, 1])}")
```
